Q1 [20 pts] Define each of the following  machine learning  terms  in your own words:

i. the training  dataset,  testing  dataset,  and validation  dataset 

The training dataset is a subset of the data used to fit the machine learning model and learn model parameters from the input features and labels.

The validation dataset is used during model development to guide decisions such as hyperparameter tuning and model selection. It is not used to fit model parameters, but instead helps evaluate different model configurations and prevent overfitting.

The testing dataset is separate, held-out portion of the data used only after all model decisions have been finalized. It provides an unbiased estimate of how well the trained model generalizes to new unseen data.

A common example ML strategy is k-fold cross validation, where the training data is split into k subsets and each fold is used as a validation set while the remaining folds are used for training. The test remains untouched until the final evaluation. 

ii. ground  truth, label 
Ground truth referes to the true, correct outcome or classification for a data point, typically obtained from reliable sources such as human annotation, measurements, or known experimental results. It represents the reference standard against which model predictions are evaluated.

A label is the specific value or category assigned to a data point that the machine learning model is trained to predict. In supervised learning, labels are derived from the ground truth and are used during training and evaluation.

iii. pre-processing,  feature, numerical, nominal 

Pre-processing refers to the steps applied to raw data before model training to make it suitable for machine learning. Common pre-processing steps include cleaning missing values, scaling or normalizing data, encoding categorical variables, and removing noise.

A feature is an individual measurable property or attribute of the data that is used as input to a machine learning model. Features are typically represented as columns in a dataset.

A numerical feature is a type of feature that takes on quantitative values with a meaningful numerical scale, like measurements or counts, and can be directly used in mathematical computations.

A nominal feature is a categorical variable with discrete labels that have no inherent ordering or numeric meaning, such as class names or categories.

iv. decision  surface 

A decision surface is the boundary in feature space that separates the different predicted classes produced by a machine learning model. It defines how the model partitions the input space and determines the predicted label for any given data point.

For simple models, the decision surface may be a linear boundary, while more complex models with more features can produce nonlinear decisions surfaces that better capture complex patterns in the data.

v. model validation,  accuracy, cross-validation 

Model validation is the process of evaluating a machine learning model during development to assess its performance and guide decisions such as model selection and hyperparameter tuning. 

Accuracy is a performance metric that measures the proportion of correct predictions made by a model out of the total number of predictions. It is commonly used for classification tasks, especially when class distributions are balanced.

Cross-validation is a resampling technique used in model validation where the dataset is repeatedly split into training and validation subsets to obtain a more reliable estiamte of model performance (similar to k-fold I mentioned earlier :) ). This ultimately helps reduce the dependence on a single train-validation split and improves robustness.

vi. parameters,  hyperparameters,  overfit 

Hyperparameters are configuration settings that are defined before training begins and control the learning process or model structure, such as the depth of a decision tree or the number of neighbors in k-nearest neighbors. They are typically tuned using validation data.

Overfitting occurs when a model learns the training ata too closely, including noise and random fluctuations, resulting in poor generalization to new, unseen data. 

vii. reclassification 

Reclassification refers to the process of assigning a new or updated class label to a data point based on the output of a trained model. This often occurs when a model is applied to new data, updated data, or when decision boundaries change due to retraining or parameter updates.

Reclassification is commonly used to evaluate how data points may shift between classes as models are improved or as additional information becomes available.
